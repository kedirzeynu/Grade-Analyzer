# Class Grade Analyzer

This notebook demonstrates a simple data processing workflow:
1. **Load data** from `class_grades.csv`.
2. **Clean missing data** using a custom Python function.
3. **Compute class averages** using **NumPy**.

```python
# Cell 1: Import required libraries
import csv
import numpy as np
```

### Step 1: Load the CSV Data

```python
# Step 1: Load the CSV file
raw_data = []

with open('class_grades.csv', mode='r') as file:
    reader = csv.reader(file)
    header = next(reader)  # Read header row: ['StudentID', 'Midterm', 'Final']
    for row in reader:
        raw_data.append(row)

print("Headers:", header)
print("\nLoaded Raw Rows:")
for row in raw_data:
    print(row)
```

### Step 2: Clean Missing Data Using a Custom Function

```python
# Step 2: Custom function to clean missing or empty grade values
def clean_score(value, default_value=0.0):
    """
    Checks if a grade value is empty or blank.
    If empty, returns default_value (0.0).
    Otherwise, converts and returns the string as a float.
    """
    if value.strip() == "":
        return default_value
    return float(value)

# Apply the cleaning function to all student records
cleaned_data = []
for row in raw_data:
    student_id = row[0]
    cleaned_midterm = clean_score(row[1], default_value=0.0)
    cleaned_final = clean_score(row[2], default_value=0.0)
    cleaned_data.append([student_id, cleaned_midterm, cleaned_final])

print("Cleaned Data [StudentID, Midterm, Final]:")
for record in cleaned_data:
    print(record)
```

### Step 3: Process the Cleaned Data Using NumPy

```python
# Step 3: Extract score columns into NumPy arrays
midterm_list = [record[1] for record in cleaned_data]
final_list = [record[2] for record in cleaned_data]

midterm_scores = np.array(midterm_list)
final_scores = np.array(final_list)

# Calculate class averages using NumPy
midterm_avg = np.mean(midterm_scores)
final_avg = np.mean(final_scores)
overall_avg = (midterm_avg + final_avg) / 2

# Print the results
print("=" * 50)
print("Class Grade Report")
print("=" * 50)
print(f"Total Students:        {len(cleaned_data)}")
print(f"Class Midterm Average: {midterm_avg:.2f}")
print(f"Class Final Average:   {final_avg:.2f}")
print(f"Overall Class Average: {overall_avg:.2f}")
print("=" * 50)
```
